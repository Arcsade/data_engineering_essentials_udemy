```python
import pandas as pd # Import pandas library
orders_columns = ['order_id', 'order_date', 'order_customer_id', 'order_status'] # Define column names
orders_df = pd.read_csv('retail_db/orders/part-00000', names=orders_columns) # Read CSV file into DataFrame
```

```python
orders_df.\
    groupby('order_status')['order_id'].\
    agg('count') # Group by order_status and count order_id
```

```python
orders_df.\
    groupby('order_status')['order_id'].\
    agg(order_count = 'count') # Group by order_status and count order_id with alias
```

```python
orders_df['order_month'] = orders_df.apply(lambda order: order.order_date[:7], axis=1)
```

```python
orders_df.head()
```

```python
orders_df.\
    groupby(['order_month','order_status'])['order_id'].\
    agg(order_count = 'count') # Group by order_month and order_status, then count order_id with alias
    
```

```python
orders_df.\
    groupby(['order_month','order_status'])['order_id'].\
    agg(order_count = 'count').\
    reset_index() # Reset index to flatten the DataFrame
```

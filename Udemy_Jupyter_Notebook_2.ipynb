```python
def sum_of_n_natural_numbers(n): # function for sum of n natural nummbers
    if n == 0:
        return 0
    else:
        return int((n * (n + 1)) / 2)
```

```python
numbers = [1, 2, 3, 4, 5]
```

```python
[sum_of_n_natural_numbers(number) for number in numbers] # list comprehension for sum of n natural numbers
```

```python
sum_of_n_natural_numbers_lambda = lambda numbers: 0 if numbers == 0 else int((numbers * (numbers + 1)) / 2) # lambda function for sum of n natural numbers
```

```python
[sum_of_n_natural_numbers_lambda(number) for number in numbers]    # list comprehension for lambda function
```

```python
list(map(sum_of_n_natural_numbers,numbers)) # map function for sum of n natural numbers
```

```python
list(map(lambda number: 0 if number == 0 else int((number * (number + 1)) / 2),numbers)) # map function for lambda function
```

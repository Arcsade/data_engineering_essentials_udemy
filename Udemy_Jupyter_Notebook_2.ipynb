```python
def sum_of_n_natural_numbers(n): # function for sum of n natural nummbers
    if n == 0:
        return 0
    else:
        return int((n * (n + 1)) / 2)
```

```python
numbers = [1, 2, 3, 4, 5]
```

```python
[sum_of_n_natural_numbers(number) for number in numbers]
```

```python
sum_of_n_natural_numbers_lambda = lambda numbers: 0 if numbers == 0 else int((numbers * (numbers + 1)) / 2) # lambda function for sum of n natural numbers
```

```python
[sum_of_n_natural_numbers_lambda(number) for number in numbers]
```
